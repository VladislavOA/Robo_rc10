```python
import time
from rc10_api.rc_api import RobotApi

robot = RobotApi(ip="10.10.10.10", read_only=False, timeout=5)

robot.motion.scale_setup.set(velocity=0.03, acceleration=0.1)
robot.controller_state.set("run")

z0 = robot._rtd_receiver.rt_data.act_tcp_x[2]

for _ in range(200):
    robot.motion.linear.jog_once("Z", "+")
    time.sleep(0.01)

z1 = robot._rtd_receiver.rt_data.act_tcp_x[2]

print("Смещение Z:", (z1 - z0) * 1000, "мм")

robot.motion.linear.jog_once_all_axis(["0", "0", "0", "0", "0", "0"])
robot._shutdown_api_sockets()
```

```python
import time
from rc10_api.rc_api import RobotApi

robot = RobotApi(ip="10.10.10.10", read_only=False, timeout=5)

try:
    if robot.safety_status.get() != "normal":
        raise RuntimeError("Safety status не normal")

    print("Payload:", robot.payload.get())

    robot.controller_state.set("run")
    robot.motion.mode.set("hold")

    print("FreeDrive включён на 10 секунд")

    end = time.monotonic() + 30

    while time.monotonic() < end:
        robot.motion.free_drive(True)
        time.sleep(0.01)  # документация требует ~100 Hz

finally:
    robot.motion.free_drive(False)
    robot.motion.mode.set("hold")
    robot._shutdown_api_sockets()

    print("FreeDrive выключен, робот в HOLD")
```
